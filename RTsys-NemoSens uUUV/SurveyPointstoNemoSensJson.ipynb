```python
import pandas as pd
import os
import json

##UPDATE##
csv_file_path = r"C:\Users\Mike.Bollinger\Documents\R\NF-24-06_TEST\Outputs\SamplingPoints\P1_StratSamp.csv" #Alex's code output folder
json_template_home = r"E:\NF2406_UUVmissionplanning" #Parent folder containing the JSON Template


# Load the CSV file
csv_data = pd.read_csv(csv_file_path)

# Display the first few rows of the CSV file to understand its structure
print(csv_data.head())



# Load the JSON template
json_template_path = os.path.join(json_template_home, "SCM_Marker_Template.json")
with open(json_template_path, 'r') as json_file:
    json_template = json.load(json_file)

# Display the JSON template structure
print(json.dumps(json_template, indent=2))
```

```python
# Define a function to create the JSON structure from the CSV data
def create_json_from_csv(csv_data, json_template):
    paths = []
    for index, row in csv_data.iterrows():
        path = {
            "name": f"Path{index + 1}",
            "stages": [
                {
                    "name": str(row['SiteID']),
                    "latitude": row['Lat'],
                    "longitude": row['Lon'],
                    "navigation": {
                        "velocity": 1,
                        "altitude": 3,
                        "waypoint": {
                            "radius": 10,
                            "timeout": 180
                        }
                    }
                }
            ]
        }
        paths.append(path)

    # Update the JSON template with the paths
    json_template['mission']['devices'][0]['paths'] = paths
    return json_template

# Generate the JSON data from the CSV data
new_json_data = create_json_from_csv(csv_data, json_template)

# Display the new JSON data structure
print(json.dumps(new_json_data, indent=2))
```

```python
from datetime import datetime

today = datetime.today()
formatted_today = today.strftime("%Y%m%d")

# Save the new JSON data to a file
new_json_file_path = os.path.join(json_template_home, formatted_today + r"MissionPoints.json")
with open(new_json_file_path, 'w') as new_json_file:
    json.dump(new_json_data, new_json_file, indent=2)

print(f"New JSON file saved to {new_json_file_path}")
```
